#### Solution is-> Structured Output

```python
from pydantic import BaseModel, Field
from typing import Optional,List
class ResponseModel(BaseModel):
    name: str = Field(description="The name of the person")
    email: str = Field(description="The email address of the person")
    age: int = Field(description="The age of the person")  
    
    
structured_llm = llm.with_structured_output(ResponseModel)   
```

```python
structured_response= structured_llm.invoke(f"please give me only name , mail, age from this text: {text}")
```

```python
class Movies(BaseModel):
    title: str = Field(description="The title of the movie")
    year: int = Field(description="The release year of the movie")
    genre: str = Field(description="The genre of the movie")
    director: str = Field(description="The director of the movie")
    cast: list[str] = Field(description="The main cast of the movie")
    rating: float = Field(description="The rating of the movie")


class AllMovies(BaseModel):
    movies: list[Movies] = Field(description="A list of movies")
```

```python
movie_llm = llm.with_structured_output(AllMovies)

response = movie_llm.invoke("Please give me 3 Sci-Fi movies with title, year, genre, director, cast and rating")

response
```
