```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-20b"
)
```

```python
response = llm.invoke("My name is Mukti.")

print(response.content)
```

```python
response = llm.invoke("What is my name?")
print(response.content)
```

```python
prompts=[
    {"role": "user", "content": "Hello, My name is Mukti."},
    {"role": "ai", "content": "Hello Mukti! 👋 How can I help you today?"},
    {"role": "user", "content": "What is my name?"},
]

response = llm.invoke(prompts)
print(response.content)
```

#### Manage history

```python
history=[]

while True:
    query = input("user: ")
    
    if query.lower() in ["exit", "quit", "bye"]:
        print("Exiting the chat. Goodbye!")
        break
    
    history.append({"role": "user", "content": query})
    
    response = llm.invoke(history)
    history.append({"role": "ai", "content": response.content})
    
    print(f"AI: {response.content}")
```
