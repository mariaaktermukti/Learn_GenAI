# Special Tokens

[CLS] → Start/classification token

[SEP] → Separates or ends sequences

[PAD] → Padding

[UNK] → Unknown token

[MASK] → Masked token


````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
````

````python
print("CLS:", tokenizer.cls_token)
print("SEP:", tokenizer.sep_token)
print("PAD:", tokenizer.pad_token)
print("UNK:", tokenizer.unk_token)
print("MASK:", tokenizer.mask_token)
````

````python
#Token IDs of Special Tokens
print("CLS ID:", tokenizer.cls_token_id)
print("SEP ID:", tokenizer.sep_token_id)
print("PAD ID:", tokenizer.pad_token_id)
print("UNK ID:", tokenizer.unk_token_id)
print("MASK ID:", tokenizer.mask_token_id)
````

````python
from transformers import AutoTokenizer

# Step 1: Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Step 2: Give input text
text = "I love AI"

# Step 3: Encode the text
encoded = tokenizer(
    text,
    return_tensors="pt"
)

# Step 4: Show Input IDs
print("Input IDs:")
print(encoded["input_ids"])

# Step 5: Show Attention Mask
print("\nAttention Mask:")
print(encoded["attention_mask"])
````

````python
texts = [
    "I love AI",
    "I love Artificial Intelligence"
]

encoded = tokenizer(
    texts,
    padding=True,
    return_tensors="pt"
)

print("Input IDs:")
print(encoded["input_ids"])

print("\nAttention Mask:")
print(encoded["attention_mask"])
````

# Build a Token Visualization Function

````python
def visualize_tokens(text, tokenizer):

    tokens = tokenizer.tokenize(text)
    token_ids = tokenizer.convert_tokens_to_ids(tokens)

    print("Text:", text)
    print("Characters:", len(text))
    print("Tokens:", len(tokens))
    print("\n" + "=" * 50)

    for i, (token, token_id) in enumerate(zip(tokens, token_ids)):
        print(
            f"{i:>2} | "
            f"Token: {token:<20} | "
            f"ID: {token_id}"
        )


text = "I am learning Generative AI"

visualize_tokens(text, tokenizer)
````

# Token Efficiency

````python
text = "I am learning Generative AI"

tokens = tokenizer.tokenize(text)

characters = len(text)
token_count = len(tokens)

efficiency = characters / token_count

print("Characters:", characters)
print("Tokens:", token_count)
print("Characters per token:", efficiency)
````
