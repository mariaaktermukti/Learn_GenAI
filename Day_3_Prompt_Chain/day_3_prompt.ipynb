# Complete chain
#### Prompt → LLM → Output Parser → transform_case

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq


# Prompt
prompts = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a translator. Translate the given text into {language}."
    ),
    (
        "human",
        "{query}"
    )
])


# LLM
llm = ChatGroq(
    model="openai/gpt-oss-20b"
)


# Output Parser
output_parser = StrOutputParser()


# Transform function
def transform_case(result):
    return result.upper()


# Complete Chain
chains = prompts | llm | output_parser | transform_case
```

```python
response = chains.invoke({
    "language": "Banglish",
    "query": "I love to play cricket."
})

print(response)
```
